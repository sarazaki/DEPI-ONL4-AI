```python
import os
import random


def thanos_project(folder_name, num):

    # create folder
    if os.path.exists(folder_name) == False:
        os.mkdir(folder_name)

    # create files
    for i in range(1, num + 1):

        name = "file" + str(i) + ".txt"
        path = folder_name + "/" + name

        file = open(path, "w")
        file.write("This is file " + str(i))
        file.close()

    # count files
    files = os.listdir(folder_name)
    print("Files before delete:", len(files))

    # delete half randomly
    half = len(files) // 2

    remove_files = random.sample(files, half)

    for f in remove_files:

        remove_path = folder_name + "/" + f
        os.remove(remove_path)

    # check result
    final = os.listdir(folder_name)

    print("Files after delete:", len(final))
    print(final)
```

```python
thanos_project("test_folder", 12)
```
