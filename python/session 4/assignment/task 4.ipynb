```python
class EvenOdd:
    """
    This class checks if a number is even or odd.
    """

    def __init__(self, number):
        """
        Constructor to store the number.
        """
        self.number = number

    def check(self):
        """
        Checks if number is even or odd.
        """
        if self.number % 2 == 0:
            return "Even"
        else:
            return "Odd"

obj = EvenOdd(7)
result = obj.check()
print(result)
```

```python
class Factorial:
    """
    This class calculates factorial of a number.
    """

    def __init__(self, number):
        """
        Store the number.
        """
        self.number = number

    def calculate(self):
        """
        Calculates factorial.
        """
        result = 1

        for i in range(1, self.number + 1):
            result = result * i

        return result

obj = Factorial(5)
answer = obj.calculate()
print(answer)
```

```python
class Palindrome:
    """
    This class checks if a word is palindrome.
    """

    def __init__(self, word):
        """
        Store the word.
        """
        self.word = word

    def check(self):
        """
        Checks palindrome.
        """
        if self.word == self.word[::-1]:
            return "Palindrome"
        else:
            return "Not Palindrome"

obj = Palindrome("level")
print(obj.check())
```

```python
class SumEven:
    """
    This class sums even numbers in a list.
    """

    def __init__(self, numbers):
        """
        Store list of numbers.
        """
        self.numbers = numbers

    def calculate(self):
        """
        Calculates sum of even numbers.
        """
        total = 0

        for num in self.numbers:
            if num % 2 == 0:
                total += num

        return total

obj = SumEven([1, 2, 3, 4, 5, 6])
print(obj.calculate())
```

```python
class ReverseList:
    """
    This class reverses a list.
    """

    def __init__(self, data):
        """
        Store list.
        """
        self.data = data

    def reverse(self):
        """
        Reverses list.
        """
        return self.data[::-1]

obj = ReverseList([1, 2, 3, 4])
print(obj.reverse())
```

```python
class Median:
    """
    This class finds the median of a list.
    """

    def __init__(self, numbers):
        """
        Store numbers list.
        """
        self.numbers = numbers

    def calculate(self):
        """
        Calculates median.
        """
        self.numbers.sort()

        n = len(self.numbers)
        mid = n // 2

        return self.numbers[mid]

obj = Median([5, 2, 7, 3, 1])
print(obj.calculate())
```

```python
class Mode:
    """
    This class finds the mode of a list.
    """

    def __init__(self, numbers):
        """
        Store numbers.
        """
        self.numbers = numbers

    def calculate(self):
        """
        Calculates mode.
        """
        max_count = 0
        mode_num = self.numbers[0]

        for num in self.numbers:
            count = self.numbers.count(num)

            if count > max_count:
                max_count = count
                mode_num = num

        return mode_num
    
obj = Mode([1, 2, 2, 3, 4, 2, 5])
print(obj.calculate())
```

```python
class Grade:
    """
    This class converts mark to grade.
    """

    def __init__(self, mark):
        """
        Store mark.
        """
        self.mark = mark

    def calculate(self):
        """
        Calculates grade.
        """
        if 85 <= self.mark <= 100:
            return "A"
        elif 75 <= self.mark < 85:
            return "B"
        elif 65 <= self.mark < 75:
            return "C"
        elif 50 <= self.mark < 65:
            return "D"
        else:
            return "F"

obj = Grade(78)
print(obj.calculate())
```

```python
class Fibonacci:
    """
    This class generates fibonacci series.
    """

    def __init__(self, num):
        """
        Store number of terms.
        """
        self.num = num

    def generate(self):
        """
        Generates fibonacci list.
        """
        series = []

        a = 0
        b = 1

        for i in range(self.num):
            series.append(a)

            c = a + b
            a = b
            b = c

        return series

obj = Fibonacci(10)
print(obj.generate())
```
