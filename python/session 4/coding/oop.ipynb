```python
from abc import ABC , abstractmethod 

class Animal(ABC):
    @abstractmethod
    def move(self):
        pass

class Bird(Animal):
    
    def move(Bird):
        print("move from Bird")
class Cat(Animal):
    
    def move(self):
        print("move from Cat")
```
