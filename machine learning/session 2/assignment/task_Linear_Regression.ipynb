```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
class LinearRegressionGD:
    def __init__(self, learning_rate=0.001, n_iters=100):
        self.lr = learning_rate
        self.n_iters = n_iters
        self.theta_0 = 0
        self.theta_1 = 0
        self.sse_history = []

    def fit(self, X, y):
        n = len(X)

        for _ in range(self.n_iters):
            y_pred = self.theta_0 + self.theta_1 * X

            error = y - y_pred

            self.theta_0 -= self.lr * (-2/n) * np.sum(error)
            self.theta_1 -= self.lr * (-2/n) * np.sum(error * X)

            sse = np.sum(error ** 2)
            self.sse_history.append(sse)

    def predict(self, X):
        return self.theta_0 + self.theta_1 * X

    def plot_training(self, X, y):
        plt.figure()
        plt.plot(self.sse_history)
        plt.title("SSE over Iterations")
        plt.show()

        plt.figure()
        plt.scatter(X, y)
        y_pred = self.predict(X)
        plt.plot(X, y_pred)
        plt.title("Regression Line")
        plt.show()
```

```python
X = np.array([50, 60, 70, 80, 90])
y = np.array([150, 180, 210, 240, 270])
```

```python
model = LinearRegressionGD(learning_rate=0.001, n_iters=100)
```

```python
model.fit(X, y)
```

```python
print("theta_0:", model.theta_0)
print("theta_1:", model.theta_1)
```

```python
prediction = model.predict(np.array([70]))
print("Predicted price:", prediction)
```

```python
model.plot_training(X, y)
```

```python
model_large = LinearRegressionGD(learning_rate=0.1, n_iters=100)
model_large.fit(X, y)
model_large.plot_training(X, y)
```

```python
model_small = LinearRegressionGD(learning_rate=0.000001, n_iters=100)
model_small.fit(X, y)
model_small.plot_training(X, y)
```
