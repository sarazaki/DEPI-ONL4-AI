```python
'''
Given an array of integers called nums
and an integer called target,
the goal is to return the indices of two numbers such that their sum equals the target.

You must not use the same element twice.
The solution should return the indices of the two numbers, and they can be returned in any order.

عندك:
ليست أرقام nums
رقم اسمه target

عايزة ترجعي أماكن (indices) رقمين مجموعهم = target
ومينفعش تستخدمي نفس العنصر مرتين.

'''

def two_sum(nums, target):
    for i in range(len(nums)):
        for j in range(i + 1, len(nums)):
            if nums[i] + nums[j] == target:
                return [i, j]



print(two_sum([2,7,11,15], 9))
print(two_sum([3,2,4], 6))
print(two_sum([3,3], 6))
```

```python
def two_sum(nums, target):
    seen = {}  # هنخزن فيه الرقم والـ index بتاعه

    for i in range(len(nums)):
        complement = target - nums[i]

        if complement in seen:
            return [seen[complement], i]

        seen[nums[i]] = i

print(two_sum([2,7,11,15], 9))
print(two_sum([3,2,4], 6))
print(two_sum([3,3], 6))
```

```python
def two_sum(nums, target):
    seen = {}

    for i in range(len(nums)):
        complement = target - nums[i]

        if complement in seen:
            return [seen[complement], i]

        seen[nums[i]] = i

    return None


nums = list(map(int, input("Enter numbers separated by space: ").split()))
target = int(input("Enter target number: "))
print(two_sum(nums, target))
```
