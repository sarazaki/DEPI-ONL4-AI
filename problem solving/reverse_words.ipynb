```python
# first solution [::-1] to reverse the list

def reverse_words(text):
    return " ".join(text.split()[::-1])


print(reverse_words("The greatest victory is that which requires no battle"))
```

```python
def reverse_words(text):
    words = text.split()            # نقسم الجملة لكلمات
    words.reverse()                 # نعكس ترتيبهم
    return " ".join(words)          # نرجعهم جملة تاني

print(reverse_words("The greatest victory is that which requires no battle"))
```

```python
# second solution using loop

def reverse_words(text):
    words = text.split()
    reversed_words = []

    for word in words:
        reversed_words.insert(0, word)  # كل مرة نحط الكلمة في الأول

    return " ".join(reversed_words)

print(reverse_words("The greatest victory is that which requires no battle"))
```

```python
# third solution without split

def reverse_words(text):
    result = ""                                  #هنخزن فيه الجملة بعد عكس ترتيب الكلمات
    word = ""                                    #هنخزن كل كلمة حرف حرف 

    for char in text:                            #هيلف على كل حرف ف الجملة 
        if char != " ":                          #لو الحرف مفيش مسافة نضيفه للكلمة
            word += char
        else:                                    #لو غير كدا يبقى خلصنا كلمة
            result = word + " " + result         
            word = ""

    result = word + " " + result                #بعد ما اللوب بيخلص لازم نضيف اخر كلمة
    return result.strip()                       #نشيل اي مسافات ف البداية والنهاية

text = "The greatest victory is that which requires no battle"
print(reverse_words(text))
```
