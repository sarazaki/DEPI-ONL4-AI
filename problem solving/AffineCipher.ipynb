```python
import math

message = input("Enter message: ")
a = int(input("Enter key a: "))
b = int(input("Enter key b: "))

if math.gcd(a, 26) != 1:
    print("Invalid key")
else:

    encrypted = ""

    for char in message:
        if char.isalpha():
            x = ord(char.upper()) - 65
            enc = (a * x + b) % 26
            encrypted += chr(enc + 65)
        else:
            encrypted += char

    print("Encrypted:", encrypted)

    # find modular inverse
    a_inv = 0
    for i in range(26):
        if (a * i) % 26 == 1:
            a_inv = i

    decrypted = ""

    for char in encrypted:
        if char.isalpha():
            y = ord(char.upper()) - 65
            dec = (a_inv * (y - b)) % 26
            decrypted += chr(dec + 65)
        else:
            decrypted += char

    print("Decrypted:", decrypted)
```
