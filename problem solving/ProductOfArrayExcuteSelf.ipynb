```python
from typing import List

class Solution:
    def productExceptSelf(self, nums: List[int]) -> List[int]:

        matrix_size = len(nums) 
        res = [1] * matrix_size
    
        prefix = 1
        for i in range(matrix_size):
            res[i] = prefix
            prefix *= nums[i]
            
        suffix = 1
        for i in range(matrix_size - 1, -1, -1):
            res[i] *= suffix
            suffix *= nums[i]
            
        return res


if __name__ == "__main__":
    sol = Solution()
  
    test1 = [1, 2, 3, 4]
    print(f"Input: {test1}")
    print(f"Output: {sol.productExceptSelf(test1)}") 

    print("-" * 30)

    test2 = [-1, 1, 0, -3, 3]
    print(f"Input: {test2}")
    print(f"Output: {sol.productExceptSelf(test2)}")
```
