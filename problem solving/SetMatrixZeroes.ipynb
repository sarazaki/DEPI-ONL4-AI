```python
class Solution:
    def setZeroes(self, matrix: list[list[int]]) -> None:
        rows = len(matrix)
        cols = len(matrix[0])
        first_row_zero = False
        first_col_zero = False

        for j in range(cols):
            if matrix[0][j] == 0: first_row_zero = True
        for i in range(rows):
            if matrix[i][0] == 0: first_col_zero = True

        for i in range(1, rows):
            for j in range(1, cols):
                if matrix[i][j] == 0:
                    matrix[i][0] = 0
                    matrix[0][j] = 0

        for i in range(1, rows):
            for j in range(1, cols):
                if matrix[i][0] == 0 or matrix[0][j] == 0:
                    matrix[i][j] = 0

        if first_row_zero:
            for j in range(cols): matrix[0][j] = 0
        if first_col_zero:
            for i in range(rows): matrix[i][0] = 0


if __name__ == "__main__":
  
    my_matrix = [
        [1, 1, 1],
        [1, 0, 1],
        [1, 1, 1]
    ]
    
    print("المصفوفة قبل التعديل:")
    for row in my_matrix:
        print(row)
    
    sol = Solution()
    sol.setZeroes(my_matrix)
    
    print("المصفوفة بعد التعديل:")
    for row in my_matrix:
        print(row)
```

```python
import copy

def setZeroesBruteForce(matrix):
    rows = len(matrix)
    cols = len(matrix[0])
    
   
    temp_matrix = copy.deepcopy(matrix)

    for r in range(rows):
        for c in range(cols):
            if matrix[r][c] == 0:
               
                for i in range(cols):
                    temp_matrix[r][i] = 0
                for i in range(rows):
                    temp_matrix[i][c] = 0
    
   
    for r in range(rows):
        for c in range(cols):
            matrix[r][c] = temp_matrix[r][c]
```

```python
def setZeroesFlags(matrix):
    rows = len(matrix)
    cols = len(matrix[0])
    
    row_flags = [False] * rows  
    col_flags = [False] * cols  


    for r in range(rows):
        for c in range(cols):
            if matrix[r][c] == 0:
                row_flags[r] = True
                col_flags[c] = True

  
    for r in range(rows):
        for c in range(cols):
            if row_flags[r] or col_flags[c]:
                matrix[r][c] = 0
```
